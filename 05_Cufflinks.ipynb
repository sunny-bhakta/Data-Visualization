```python
import seaborn as sns
import plotly.express as px
import pandas as pd

# Load dataset
tips = sns.load_dataset('tips')
tips.head()
```

```python
fig = px.scatter_matrix(
    tips,
    # x='total_bill',
    # nbins=20,                   # number of bins
    # title='Total Bill Distribution',
    # labels={'total_bill':'Total Bill ($)'},
    # color_discrete_sequence=['blue']  # set color
)
fig.show()
```

```python
fig = px.histogram(
    tips,
    x='total_bill',
    # nbins=20,                   # number of bins
    # title='Total Bill Distribution',
    # labels={'total_bill':'Total Bill ($)'},
    # color_discrete_sequence=['blue']  # set color
)
fig.show()
```

```python
fig = px.scatter_matrix(
    tips,
    dimensions=['total_bill', 'tip', 'size'],
    color='sex',
    title='Scatter Matrix of Tips Dataset'
)
fig.show()
```

```python
fig = px.bar(
    tips.groupby('day')['tip'].mean()
)
fig.show()
```
